# Задача 1: угловая
## Дано:
Звезды из файла `stars_main.csv`. Не учитывать колонку `epoch` и считать, что координаты всех звезд даны **на одну эпоху.**
## Используя:
* $\alpha$ — прямое восхождение звезды;
* $\delta$ — склонение звезды;

## Найти:
* Наибольшее и наименьшее угловое расстояние между парами из этих звезд.

## Необходимый минимум

* Перевод прямого восхождения из часовой меры (часы, минуты, секунды) или градусов в радианы.
* Сферическая тригонометрия: угловое расстояние между двумя точками на сфере (теорема косинусов для сферы или формула гаверсинусов / формула Винсенти для устранения погрешностей при малых углах).

# Задача 4: скоростная
## Дано: 
Звезды из файла `stars_main.csv`. Не учитывать колонку `epoch` и считать, что координаты и скорости всех звезд даны **на одну эпоху.**

## Используя:
* $\alpha$ — прямое восхождение звезды;
* $\delta$ — склонение звезды;
* $\varpi$ — годичный параллакс звезды; 
* $\mu_\alpha$ — собственное движение звезды по $\alpha$;
* $\mu_\delta$ — собственное движение звезды по $\delta$;
* $v_R$ — барицентрическая (ICRS) лучевая скорость звезды;

## Найти:
* Три компоненты вектора скорости $\vec{v}=(v_x, v_y, v_z)$ в ICRS для каждой звезды.

## Необходимый минимум
* Понятие годичного параллакса $\varpi$ и связь параллакса с расстоянием до звезды в парсеках. Величина парсека в а.е. и км.
* Собственное движение звезды ($\mu_\alpha$, $\mu_\delta$) и перевод угловой скорости в компоненты тангенциальной скорости ($v_\alpha$, $v_\delta$) в км/с через расстояние.
* Соотношение между $1\text{ mas/yr}$ и физической скоростью на заданной дистанции.
* Переход от сферических координат в ICRS ($\alpha$, $\delta$) и компонент скорости ($v_\alpha$, $v_\delta$, $v_R$) к декартовым компонентам вектора скорости $(v_x, v_y, v_z)$.

# Задача 7*: солнечная
## Дано: 
Файл `sun_locations.json`.

## Используя:
* Широту и долготу пункта на Земле;
* Смещение его часовой зоны относительно нулевого меридиана;
* Дату в этом пункте;

## Учитывая:
* Постоянные размеры диска Солнца $r_\odot=16^\prime$;
* Постоянное значение атмосферной рефракции на горизонте $\rho = 35^\prime$;
* Вариацию уравнения времени $\eta$ — разницы между средним и истинным солнечным временем;

## Считая фиксированным:
* Одно из мартовских равноденствий происходит ровно **20.03.2042 T=12:00** нулевого меридиана;
* Величина тропического года **ровно** $T_0 = 365.2425~$суток;
* Средние солнечные сутки равны **ровно** $86400$ секунд;
* Среднее солнечное время постоянно.
    
## Найти: 
* местное время восхода Солнца в дату X с точностью до минут,
* местное время захода Солнца в дату X с точностью до минут.

## Необходимый минимум
* Горизонтальная система координат и экваториальная система координат.
* Понятие часового угла ($t$), склонения Солнца ($\delta_\odot$) и широты наблюдателя ($\varphi$).
* Изменение эклиптической долготы Солнца от даты относительно равноденствия.
* Определение момента восхода и момента захода Солнца.
* Понятия верха/центра диска Солнца и влияние видимого радиуса $r_\odot$ и атмосферной рефракции $\rho$ на зенитное расстояние при восходе/заходе
* Связь истинного солнечного времени, среднего солнечного времени и местного гражданского времени.

# Задача 8*: кеплеровая
## Дано:
Файл `orbital_params.json`, содержащий разные параметры орбиты малого тела, которое обращается вокруг центрального тела с гравитационным параметром $GM = 39.478$ а.е.³/год².

## Используя:
* Большую полуось орбиты;
* Эксцентриситет орбиты;

## Построить:
* Оси координат $x$ и $y$, где начало отсчета совпадает с центральным телом, ось $x$ направлена в перицентр орбиты;
* Плоскую эллиптическую орбиту этого тела в этих координатах в виде 20 точек, причем кеплерово движение этого тела между любой соседней парой точек длится **одно и то же время** $\Delta T$;

## Найти:
Величину этого времени $\Delta T$ в годах.

## Необходимый минимум
* Законы Кеплера и геометрия эллиптической орбиты (большая полуось $a$, эксцентриситет $e$, фокусы).
* Орбитальные аномалии: средняя аномалия ($M$), эксцентрическая аномалия ($E$), истинная аномалия ($\nu$).
* Уравнение Кеплера ($M = E - e \sin E$).
* Численные методы решения трансцендентного уравнения Кеплера (метод Ньютона-Рафсона или метод простой итерации).
* Вычисление плоских декартовых координат $(x, y)$ планеты через эксцентрическую или истинную аномалию.  

# Описание входных файлов
## 1 — `stars_main.csv`
Разделенная запятыми табличка с параметрами звезд для задач **«Угловая», «Скоростная» и «Небесная».**
### Описание колонок
* alpha_deg — прямое восхождение звезды в градусах;
* delta_deg — склонение звезды в градусах;
* pm_alpha_mas_per_yr — собственное движение звезды по прямому восхождению в миллисекундах дуги (мсд) в год (поправленное за $\cos \delta$),
* pm_delta_mas_per_yr — собственное движение звезды по склонению в мсд/год;
* parallax_mas — годичный параллакс звезды в мсд;
* rad_vel_km_per_s — барицентрическая лучевая скорость звезды в км/с;
* Vmag — видимая звездная величина;
* epoch — эпоха координат alpha и delta.

## 2 — `magnitudes.csv`
Колонка с звездными величинами набора звезд для задачи **«Величинная»**
### Описание колонок
* magnitude — видимая звездная величина.

## 3 — `telescopes.csv`
Разделенная запятыми табличка с параметрами телескопа для задачи **«Телескопная»**
### Описание колонок
* aperture_mm — диаметр апертуры телескопа в мм;
* magnification — увеличение телескопа;
* mag_eye — предельная звездная величина звезд, видимых невооруженным глазом.

## 4 — `coord_tranforms.json`
JSON-табличка со списком словарей входных данных для задачи **«Координатная»**
### Описание ключей
* system_type — тип координатной системы:
  * AltAz — горизонтальная система, (южный) азимут и высота;
  * Equ — II экваториальная система, прямое восхождение и склонение;
  * Ecl — эклиптическая система, эклиптическая долгота и широта;
  * Gal — галактическая система, галактическая долгота и широта.
* latitute_deg — астрономическая широта места наблюдения в градусах;
* longitude_deg — астрономическая долгота места наблюдения в градусах;
* ERA_deg — угол вращения Земли на момент наблюдения в градусах;
* coord_1_deg — значение угловой абсциссы в градусах (азимут, прямое восхождение, долгота)
* coord_2_deg — значение угловой ординаты в градусах (высота, склонение, широта)

## 5 — `sun_locations.json`
JSON-табличка со списком словарей входных данных для задачи **«Солнечная»**
### Описание ключей
* description — описание пункта и даты;
* latitute_deg — географическая широта места наблюдения в градусах;
* longitude_deg — географическая долгота места наблюдения в градусах;
* timezone_hours — номер часовой зоны, или разница местного гражданского времени и всемирного;
* date_iso — дата пункта в формате ГОД-МЕСЯЦ-ДЕНЬ.

## 6 — `orbital_params.json`
JSON-табличка со списком словарей входных данных для задачи **«Кеплеровая»**
### Описание ключей
* description — описание орбиты;
* a_au — размер большой полуоси в астрономических единицах;
* e — эксцентриситет орбиты.
